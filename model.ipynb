```python
numerical_col=df.select_dtypes(include=("int64","float"))
numerical_col
categorical=df.select_dtypes(exclude=("int64","float"))
categorical


x=df.drop(columns=['Failure_Status','Timestamp','Machine_ID'])
y=df['Failure_Status']


#• Training Data = The Homework & Practice Problems. The student (the model) studies these questions along with the answer key to learn the subject.
# Test Data = The Final Exam. The teacher (the engineernumerical_col=df.select_dtypes(include=("int64","float"))
numerical_col
categorical=df.select_dtypes(exclude=("int64","float"))
categorical


x=df.drop(columns=['Failure_Status','Timestamp','Machine_ID'])
y=df['Failure_Status']


#• Training Data = The Homework & Practice Problems. The student (the model) studies these questions along with the answer key to learn the subject.
# Test Data = The Final Exam. The teacher (the engineer) g) gives the student brand-new questions they haven't seen before. If the student scoring 100% on homework gets a 40% on the final exam, they just memorized the homework without actually learning the subject.


x_train, xtest, y_train, ytest = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)


pipe=Pipeline([
    ('imputer',SimpleImputer(strategy='median')),#imputer is used for missing values 
    ('scaler',StandardScaler()),
    ('model',LogisticRegression(
        max_iter=100,
        random_state=42,
        class_weight='balanced'

    ))
])

pipe.fit(x_train,y_train)
pred=pipe.predict(xtest)

confusion_matrix_0=confusion_matrix(ytest,pred)
confusion_matrix_0 #array([[486,  54], [ 55,   5]])


classification_0=classification_report(ytest,pred)
print(classification_0)
```

```python
pipe1=Pipeline([
    ('imputer',SimpleImputer(strategy='median')),#imputer is used for missing values 
    ('model',DecisionTreeClassifier(
        random_state=42,
        class_weight='balanced',
        min_samples_leaf=10,
        max_depth=5 ))#it can keep splitting until it learns overly specific patterns.If we don't control the depth of a Decision Tree, it can grow too complex and overfit the training data. It may memorize noise and specific patterns instead of learning relationships that generalize to unseen data. We can reduce overfitting using parameters such as max_depth and min_samples_leaf, then evaluate the model on unseen data
])

pipe1.fit(x_train,y_train)
ypred_on_pipe1=pipe1.predict(xtest)
ypred_on_pipe1


confusion_matrix1=confusion_matrix(ytest,ypred_on_pipe1)
confusion_matrix1 #array([[486,  54], [ 55,   5]])


classification1=classification_report(ytest,ypred_on_pipe1)
print(classification1)
```

```python
pipe3=Pipeline([
    ('imputer3',SimpleImputer(strategy='median')),
    ('mosel3',GradientBoostingClassifier(
        n_estimators=100,
        random_state=42,
        max_depth=3,
        learning_rate=0.05       
    ))
])
pipe3.fit(x_train,y_train)
ypred_on_pipe3=pipe3.predict(xtest)
ypred_on_pipe3

confusion_matrix3=confusion_matrix(ytest,ypred_on_pipe3)
confusion_matrix3 #


classification3=classification_report(ytest,ypred_on_pipe3,zero_division=0)
print(classification3)
```

```python
pipe2=Pipeline([
    ('imputer',SimpleImputer(strategy='median')),#imputer is used for missing values 
    ('model',RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight='balanced',
        min_samples_leaf=10,
        max_depth=8 ))#it can keep splitting until it learns overly specific patterns.If we don't control the depth of a Decision Tree, it can grow too complex and overfit the training data. It may memorize noise and specific patterns instead of learning relationships that generalize to unseen data. We can reduce overfitting using parameters such as max_depth and min_samples_leaf, then evaluate the model on unseen data
])

pipe2.fit(x_train,y_train)
ypred_on_pipe2=pipe2.predict(xtest)
ypred_on_pipe2


accuracy_score2=accuracy_score(ytest,ypred_on_pipe2) #overall prediced pecentage that your model got right 
accuracy_score2#0.81

precision_score2=precision_score(ytest,ypred_on_pipe2)#how many positive prediction are actually correct
precision_score2#0.084

recall2=recall_score(ytest,ypred_on_pipe2)# how many are actual positive cases that your model succesfully found
recall2 #0.0833,,0.13

f1_score2=f1_score(ytest,ypred_on_pipe2)
f1_score2 #0.084

confusion_matrix2=confusion_matrix(ytest,ypred_on_pipe2)
confusion_matrix2 #array([[486,  54], [ 55,   5]])


classification2=classification_report(ytest,ypred_on_pipe2)
print(classification2)

prob=pipe3.predict_proba(xtest)
```

```python
x_fit,x_val,y_fit,y_val = train_test_split(x_train,y_train, test_size=0.20, random_state=42, stratify=y_train)

print([x_fit.shape,x_val.shape])
print([y_fit.shape,y_val.shape])

print(y_val.value_counts())

print(y_fit.value_counts())

print(y_fit.value_counts(normalize=True).round(3))
print(y_val.value_counts(normalize=True).round(3))



pipe4=Pipeline([
        ('imputer4',SimpleImputer(strategy='median')),
        ('model4',GradientBoostingClassifier(
            n_estimators=100,
            random_state=42,
            max_depth=3,
            learning_rate=0.05      
        ))
    ])


pipe4.fit(x_fit,y_fit)
prob4=pipe4.predict_proba(x_val)[:,1]

thres1 = [0.05, 0.10, 0.15, 0.20, 0.30, 0.40, 0.50]

for i in thres1:
    pred4 = (prob4 >= i).astype(int)

    print(
        f"Threshold: {i:.2f} | "
        f"Precision: {precision_score(y_val, pred4, zero_division=0):.3f} | "
        f"Recall: {recall_score(y_val, pred4, zero_division=0):.3f} | "
        f"F1: {f1_score(y_val, pred4, zero_division=0):.3f}|"      
    )
```

```python
#  model comparison 
model={
    "LogisticRegression":pipe,
    "DecisionTreeClassifier":pipe1,
    "RandomForestClassifier":pipe3,
    "GradientBoostingClassifier":pipe4

}

for name, models in model.items():
    prob_m=models.predict_proba(x_val)[:,1]
    pred_m=(prob_m>=0.5).astype(int)
    print(f"\n{name}")
    print("Precision:", round(precision_score(y_val, pred_m, zero_division=0),3))
    print("Recall:", round(recall_score(y_val, pred_m, zero_division=0),3))
    print("F1:", round(f1_score(y_val, pred_m, zero_division=0),3))#  model 
```

```python
dt_prob=pipe1.predict_proba(x_val)[:,1]
for thres in [0.20, 0.30, 0.40, 0.50, 0.60]:
    dt_pred=(dt_prob>=thres).astype(int)#converting into int format

    print(
        f"Threshold: {thres:.2f} | ",#2d floating value
        f"precision:{precision_score(y_val,dt_pred, zero_division=0):.3f}|",#3d floating value
        f"recall:{recall_score(y_val,dt_pred, zero_division=0):.3f}|",
        f"f1:{f1_score(y_val,dt_pred, zero_division=0):.3f}|"


    )
```
