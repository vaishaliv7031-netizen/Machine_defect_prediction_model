```python
import uvicorn
import asyncio
import requests
```

```python
from main import app

config = uvicorn.Config(
    app,
    host="127.0.0.1",
    port=8000,
    log_level="info",
)
```

```python
server =uvicorn.Server(config)#create the server

server_task=asyncio.create_task(server.serve())#schedules the server in the background  and then start listning http requests
```

```python
#creating the server status 
await asyncio.sleep(1)
print("server strated ", server.started)
```

```python
print("weather the server_tasks is inished ?:",server_task.done())
```

```python
response = await asyncio.to_thread(requests.get, "http://127.0.0.1:8000/docs")
print(response.status_code)
```
